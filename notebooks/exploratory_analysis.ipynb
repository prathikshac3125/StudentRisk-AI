# StudentRisk AI — Exploratory Analysis
This notebook explores the UCI Student Performance dataset. Run from the project root.

```python
from src.load_data import load_student_data
import matplotlib.pyplot as plt
import pandas as pd
df = load_student_data()
display(df.head())
print('Shape:', df.shape)
display(df[['G1', 'G2', 'G3', 'absences', 'studytime']].describe())
```

```python
df['at_risk'] = (df['G3'] < 10).astype(int)
df['at_risk'].value_counts().rename(index={0: 'Not identified as at risk', 1: 'At risk'}).plot(kind='bar')
plt.title('Student counts by project risk label')
plt.ylabel('Number of students')
plt.tight_layout()
plt.show()
```
